```python
# Databricks notebook source

# MAGIC %md
# MAGIC # PySpark - RDD and DataFrame
# MAGIC
# MAGIC **RDD (Resilient Distributed Dataset)** is an immutable,
# MAGIC distributed collection of objects.
# MAGIC
# MAGIC In modern Databricks environments using Spark Connect,
# MAGIC we use the existing `spark` session instead of creating
# MAGIC a new `SparkContext`.

# COMMAND ----------

# DBTITLE 1,Create Sample Data

data = [
    {"Name": "Arun", "ID": 10, "City": "KOL"},
    {"Name": "Barun", "ID": 12, "City": "PUNE"},
    {"Name": "Pabitra", "ID": 3, "City": "CHEN"},
    {"Name": "Esha", "ID": 14, "City": "MUMBAI"}
]
```

```python
# COMMAND ----------

# DBTITLE 1,Create DataFrame from List of Dictionaries

df = spark.createDataFrame(data)
df.show()
```

```python
# COMMAND ----------

# DBTITLE 1,Display Schema

df.printSchema()
```

```python
# COMMAND ----------

# DBTITLE 1,Display Columns

df.columns

```

```python
# COMMAND ----------

# DBTITLE 1,Display Number of Records

df.count()
```

```python
#COMMAND
# DBTITLE 1,Create DataFrame from List

sampledata = [
    ("Pabitra Kumar", "CTS", "KOL"),
    ("Malathi G", "CTS", "CKC"),
    ("Venkatesh V", "CTS", "CKC"),
    ("Yatharth Puri", "IBM", "LON"),
    ("Vignesh S", "CTS", "UK"),
    ("Rambabu V", "CTS", "BLR"),
    ("Bikram Jena", "TCS", "UK")
]

# COMMAND ----------

# DBTITLE 1,Define Column Names

columns = [
    "Employee_Name",
    "Company",
    "Location"
]
```

```python
# COMMAND ----------

# DBTITLE 1,Create DataFrame

df_employee = spark.createDataFrame(
    data=sampledata,
    schema=columns
)
```

```python
# COMMAND ----------

# DBTITLE 1,Display DataFrame

df_employee.show()
```

```python
# COMMAND ----------

# DBTITLE 1,Display Schema

df_employee.printSchema()
```

```python
# COMMAND ----------

# DBTITLE 1,Select Columns

df_employee.select(
    "Employee_Name",
    "Company",
    "Location"
).show()
```

```python
# COMMAND ----------

# DBTITLE 1,Select Specific Columns

df_employee.select(
    "Employee_Name",
    "Company"
).show()
```

```python
# COMMAND ----------

# DBTITLE 1,Filter Data

df_employee.filter(
    df_employee.Company == "CTS"
).show()
```

```python
# COMMAND ----------

# DBTITLE 1,Filter Employees from UK

df_employee.filter(
    df_employee.Location == "UK"
).show()
```

```python
# COMMAND ----------

# DBTITLE 1,Count Records

df_employee.count()
```

```python
# COMMAND ----------

# DBTITLE 1,Display Data Using Databricks display()

display(df_employee)
```

```python
# COMMAND ----------

# DBTITLE 1,Create DataFrame with Explicit Schema

from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    IntegerType
)
```

```python
# COMMAND ----------

# DBTITLE 1,Define Schema

schema = StructType([
    StructField("Name", StringType(), True),
    StructField("ID", IntegerType(), True),
    StructField("City", StringType(), True)
])
```

```python
# COMMAND ----------

# DBTITLE 1,Create DataFrame Using Explicit Schema

df_schema = spark.createDataFrame(
    data=data,
    schema=schema
)
```

```python
# COMMAND ----------

# DBTITLE 1,Display DataFrame

df_schema.show()
```

```python
# COMMAND ----------

# DBTITLE 1,Display Schema

df_schema.printSchema()
```
