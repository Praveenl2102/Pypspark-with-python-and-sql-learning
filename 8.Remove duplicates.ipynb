```python
# Databricks notebook source

from pyspark.sql import SparkSession
from pyspark.sql.functions import *

# Create Spark Session
spark = SparkSession.builder \
    .appName("DataFrame") \
    .getOrCreate()

# COMMAND ----------

# Read IPL CSV file

df_1 = spark.read \
    .format("csv") \
    .option("inferSchema", "true") \
    .option("header", "true") \
    .load("/Volumes/workspace/default/ipldata/ipldata.csv")

# Display DataFrame
df_1.show()

# Count total records
df_1.count()

# COMMAND ----------

# Remove duplicate rows using distinct()

df_2 = df_1.distinct()

# Display distinct records
df_2.show()

# Count distinct records
df_2.count()

# COMMAND ----------

# Remove duplicates based on a specific column


df_3 = df_1.dropDuplicates(["match_id"])

# Display result
df_3.show()

# Count records
df_3.count()

# COMMAND ----------
```
