```python
# Databricks notebook source

from pyspark.sql import SparkSession
from pyspark.sql.functions import *

# Create Spark Session
spark = SparkSession.builder \
    .appName("DataFrame") \
    .getOrCreate()

# COMMAND ----------

# Read IPL CSV file
df_1 = spark.read \
    .format("csv") \
    .option("inferSchema", "true") \
    .option("header", "true") \
    .load("/Volumes/workspace/default/ipldata/ipldata.csv")

# Display first 4 records
df_1.show(4)

# COMMAND ----------

# Add new columns
df_2 = df_1 \
    .withColumn("Country", lit("INDIA")) \
    .withColumn("Org_Id", concat(col("match_id"), col("season")))

# Display first 10 records
df_2.show(10)

# COMMAND ----------
```
