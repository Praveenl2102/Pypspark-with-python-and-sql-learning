```python
%sql
select * from employees;
```

```python
# Read employees table
df1 = spark.table("workspace.default.employees")

# Check data
df1.show()

# Check schema
df1.printSchema()

# COMMAND ----------

# Convert Salary to numeric
df1 = df1.withColumn(
    "Salary",
    col("Salary").cast("double")
)

# COMMAND ----------

# Group by Department
df3 = df1.groupBy("department") \
    .agg(
        sum("Salary").alias("Total_Salary"),
        max("Salary").alias("Max_Salary"),
        min("Salary").alias("Min_Salary")
    )

# Display result
df3.show()
```
