```python
%python
data = [10, 20, 30, 40, 50]

df = spark.createDataFrame(
    [(x,) for x in data],
    ["number"]
)

display(df)
```

```python
%python
data = [
    (1, "Praveen", "Hyderabad", 50000),
    (2, "Rahul", "Bangalore", 60000),
    (3, "Anil", "Chennai", 45000),
    (4, "Priya", "Hyderabad", 70000)
]

columns = [
    "id",
    "name",
    "city",
    "salary"
]

df = spark.createDataFrame(data, columns)

display(df)
```

```python
%python
df.select("name", "salary").display()
```

```python
%python
df.filter(df.salary > 50000).display()
```

```python
%python
df.orderBy(df.salary.desc()).display()
```

```python
%python
df.count()
```

```python
%python
df.groupBy("city").count().display()
```

```python
%python
data = [
    (1, "Praveen", "IT", 50000),
    (2, "Rahul", "HR", 45000),
    (3, "Anil", "IT", 60000),
    (4, "Priya", "Finance", 70000)
]

columns = [
    "employee_id",
    "employee_name",
    "department",
    "salary"
]

df = spark.createDataFrame(data, columns)
```

```python
df.write.mode("overwrite").saveAsTable(
    "workspace.default.employees"
)
```
